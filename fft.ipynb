```python
import numpy as np
import cupy as cp
import matplotlib.pyplot as plt
import math

x_values = np.linspace(0, 1, 1000)
y_values = [math.sin(x) for x in x_values]
y2_values = np.fft.rfft(y_values)
print(y2_values)
print(max(y2_values))
plt.plot(x_values, y2_values)
plt.plot(x_values, y_values)
plt.show()
```
